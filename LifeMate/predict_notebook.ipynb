```python
# Set the same fixed seed as the prediction command.
from pathlib import Path
from lifemate.config import load_config
from lifemate.workflows import seed_inference, load_prediction_models, prediction_inputs, predict

config_path = Path('configs/inference.competition.json')
config = load_config(config_path, 'inference', {'out': '/result/jupyter_submission.jsonl'})
seed_inference()
```

```python
# Load both local models once.
models = load_prediction_models(config)
print(f'Model load: {models.load_seconds:.3f} seconds')
```

```python
# Read the test manifest and split catalog. Audio is read inside per-sample timing.
inputs = prediction_inputs(config)
print(f'{len(inputs[1])} samples in {config.split}')
```

```python
# Run the shared agent and publish the independently generated timed submission.
try:
    summary = predict(config, models=models, prepared_inputs=inputs)
finally:
    models.close()
print(f"Prediction: {summary['prediction_seconds']:.3f} seconds")
print(summary['output'])
```
